```python
# Visualize the survival rate of passengers based on their ticket class (1st, 2nd, or 3rd class). Were the passengers in higher classes more likely to survive?

# Step 1: Imports
import pandas as pd 
import matplotlib.pyplot as plt

# Step 2: Load the Internet 
df = pd.read_csv(r"Titanic.csv")
```

```python
df.head()
```

```python
# Step 3: Group the data average of "survival" because 0/1: mean = survival rate 

class_survival = df.groupby("class")["survived"].mean()

# Step 4: Plot survival rate by the passenger class 
plt.bar(class_survival.index, class_survival.values, color=["#201C52", "#00DBA8", "#02E6ED"])
plt.title("Survival Rate by Passanger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# Create a histogram to visualize the distribution of passenger ages. What is the age distribution like on the Titanic?

plt.hist(df["age"].dropna(), bins=20, color="#00BDA8", edgecolor="black")
plt.title("Age Distribution of Passenger")
plt.xlabel("Age")
plt.ylabel("Frequancy")
plt.show()
```

```python
# Generate a bar chart to compare the survival rates of male and female passengers. Did gender play a role in survival?

gender_survival = df.groupby("sex")["survived"].mean()
plt.bar(gender_survival.index, gender_survival.values, color=["#201C52", "#00BDA8"])
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# Visualize the distribution of ticket fares for each passenger class using box plots. How did ticket fares vary across different classes?

df.boxplot(column="fare", by="class", grid=False)
plt.title("Ticket fare distribution by class")
plt.subtitle(" ")
plt.xlabel("Passanger CLass") 
plt.ylable("Fare")
plt.show()
```
